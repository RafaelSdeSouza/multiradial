# MultiRadial tutorial

Build support-constrained geometry once, then apply it to registered tracers. The support and centres—not the measured intensity—define radial position.

```python
%pip install -q "multiradial[demo]"
from multiradial import build_geometry, radial_profile
from multiradial.plotting import plot_overview
from multiradial.synthetic import make_scene
```

```python
scene = make_scene("folded")
geometry = build_geometry(scene.support, scene.centres)
geometry.extents
```

`rho_D` describes relative centre–boundary depth. `rho_X` describes normalized progression to the furthest point in each centre-associated region.

```python
brightness_D = radial_profile(scene.brightness, geometry, coordinate="rho_D")
brightness_X = radial_profile(scene.brightness, geometry, coordinate="rho_X")
colour_X = radial_profile(scene.tracer, geometry, coordinate="rho_X")
brightness_D.median.shape, colour_X.median.shape
```

```python
figure, axes = plot_overview(
    scene.brightness, geometry, coordinate="rho_X", profile=brightness_X
)
```

The default estimator uses 30 equal bins on `[0, 1]`, includes `rho=1` in the final bin, omits bins with fewer than six pixels, and reports unweighted medians with 16th/84th percentiles. No spline is fitted.
