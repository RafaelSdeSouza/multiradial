# Interactive MultiRadial explorer

Use the controls to compare support geometry, coordinate choice, and registered tracers.

```python
%pip install -q "multiradial[demo]"
import matplotlib.pyplot as plt
import panel as pn
from multiradial import build_geometry, radial_profile
from multiradial.plotting import plot_overview
from multiradial.synthetic import make_scene
pn.extension()
```

```python
shape = pn.widgets.Select(name="Support", options=["compact", "folded", "perforated", "branched"], value="folded")
coordinate = pn.widgets.RadioButtonGroup(name="Coordinate", options=["rho_D", "rho_X"], value="rho_D")
tracer = pn.widgets.RadioButtonGroup(name="Tracer", options=["brightness", "tracer"], value="brightness")
```

```python
@pn.depends(shape, coordinate, tracer)
def view(shape, coordinate, tracer):
    scene = make_scene(shape)
    geometry = build_geometry(scene.support, scene.centres)
    field = getattr(scene, tracer)
    profile = radial_profile(field, geometry, coordinate=coordinate)
    figure, _ = plot_overview(field, geometry, coordinate=coordinate, profile=profile)
    pane = pn.pane.Matplotlib(figure, format="svg", tight=True)
    plt.close(figure)
    return pane

pn.Column(pn.Row(shape, coordinate, tracer), view).servable()
```
